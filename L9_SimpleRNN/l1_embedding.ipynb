```python
from tensorflow.keras.preprocessing.text import one_hot
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Input
```

```python
sent = [
'I drink milk every morning',
'I drink water during the day',
'I brush my teeth after waking up',
'eat breakfast before work',
'I take a short walk outside',
'I cook dinner in the evening',
'clean my room every weekend',
'I read a book before sleeping',
'go grocery shopping on Sunday',
'I listen to music while cooking',
]
```

```python
vocab_size = 10000
one_hot_rep = [one_hot(words,vocab_size) for words in sent]
```

```python
one_hot_rep
```

```python
sents = pad_sequences(one_hot_rep, padding='pre', maxlen=8)
sents
```

```python
# each sentence is a array of length 8, now each word/int in the sent will be converted to a vector of length 10, 
# so the final shape will be (8,10) where 8 is the number of words in each sentence and 10 is the length of the vector for each word
dim = 10

model = Sequential()
model.add(Input(shape=(8,))) # input_shape parameter in below embedding layer is deprecated
model.add(Embedding(vocab_size, dim))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
model.predict(sents)[0]
```

# Word2vec concept elaborated below
1. Word2vec has a 300 dimension, 1 billion words dataset by google. We can use it to generate a vector for any word and it is an array with 300 ints.
2. CBOW (contionous BOW) and SkipGram are frameworks used above this word2vec.
3. CBOW and SG are opposite in nature. With CBOW you can use sent to get a word whereas in SG you can use a word to get a sent

```python
# install gensim
from gensim.models import word2vec, keyedvectors
import gensim.downloader as api
```

```python
# dont run everytime, takes time to download the model
wv = api.load('word2vec-google-news-300')
```

```python
vec_king = wv['king']
print('vector size: ', vec_king.shape)
print('vector:', vec_king)
```

```python
# most similar word present in corpus which has similar vec to king
wv.most_similar('king')
```

```python
# find similarity between 2 words
print(wv.similarity('boy','girl'))
print(wv.similarity('football','sport'))
```

```python
# vec adding and substraction
vec_calcu = wv['king'] - wv['man'] + wv['woman']

wv.most_similar(vec_calcu)
```

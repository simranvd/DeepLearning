1. Reloading input feature engineering and applying it on new input data
2. Reloading model and predicting output based on the model trained on new input

```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
loaded_model = load_model('churn_model.h5')
```

```python
with open('scaler.pkl', 'rb') as f:   #read byte - rb, write byte - wb
    scaler = pickle.load(f)

with open('gender_encoder.pkl', 'rb') as f:
    gender_encoder = pickle.load(f)

with open('geography_encoder.pkl', 'rb') as f:
    geography_encoder = pickle.load(f)
```

```python
input_data = pd.DataFrame([{'CreditScore':600,
                           'Geography':'France',
                           'Gender':'Male',
                           'Age':40,
                           'Tenure':5,
                           'Balance':60000,
                           'NumOfProducts':2,
                           'HasCrCard':1,
                           'IsActiveMember':1,
                           'EstimatedSalary':50000,
}])

input_data
```

```python
input_data['Gender'] = gender_encoder.transform(input_data['Gender'])
geography_new = pd.DataFrame(geography_encoder.transform(input_data[['Geography']]),columns=geography_encoder.get_feature_names_out(['Geography']))
```

```python
input_data = pd.concat([input_data.drop('Geography', axis=1),geography_new],axis=1)
input_data.head()
```

```python
X = scaler.transform(input_data)
```

```python
# prediction
prediction = loaded_model.predict(X)
print(prediction)
```

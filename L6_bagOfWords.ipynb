```python
import pandas as pd

mydata = pd.read_csv('requirements/emailClassify.csv',sep='\t',header=None,names=['output','message'])
print(mydata)
```

```python
# Data Cleaning and Preprocessing
import nltk
from nltk.corpus import stopwords
import re

lemm = nltk.stem.WordNetLemmatizer()
stopwords = set(stopwords.words('english'))
corpus = []

for i in range(len(mydata)):
    #remove special characters and digits
    msg = re.sub(r'[^a-zA-Z]',' ',mydata['message'][i])

    #convert to lowercase
    msg=msg.lower()

    #tokenization
    msg = nltk.word_tokenize(msg)

    #lemmatize and remove stop words
    msg = [lemm.lemmatize(word, pos='v') for word in msg if word not in stopwords]
    msg = ' '.join(msg)

    #add messages to corpus
    corpus.append(msg)

print(corpus)
```

## BOW

```python
# Bag of Words Model to convet text to feature vectors
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=100)
# cv = CountVectorizer(max_features=100, binary=True) --> for binary features
X = cv.fit_transform(corpus).toarray()

print(X)
```

## BOW with N-gram

```python
cv = CountVectorizer(max_features=300, ngram_range=(1,2)) # range=(2,3) for bigrams and trigrams
X = cv.fit_transform(corpus).toarray()

cv.vocabulary_  # to see the features generated by the model
```

## TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_features=100)  # Here you can also use ngram_range=(1,2) for bigrams and trigrams
X = tfidf.fit_transform(corpus).toarray()

print(X)
```
